```python
# ============================================
# DATA QUALITY CHECKS
# ============================================

print("========== DATA QUALITY CHECK ==========\n")

# 1. Check duplicate Ticket IDs
duplicate_ticket_ids = df["Ticket ID"].duplicated().sum()

print(f"Duplicate Ticket IDs: {duplicate_ticket_ids}")

# 2. Check unique values in important categorical columns
categorical_columns = [
    "Customer Gender",
    "Ticket Type",
    "Ticket Status",
    "Ticket Priority",
    "Ticket Channel"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts(dropna=False))

# 3. Check Customer Age validity
invalid_age = df[
    (df["Customer Age"] < 18) |
    (df["Customer Age"] > 100)
]

print("\nInvalid Customer Ages:", len(invalid_age))

# 4. Check Customer Satisfaction Rating validity
invalid_rating = df[
    (~df["Customer Satisfaction Rating"].isna()) &
    (
        (df["Customer Satisfaction Rating"] < 1) |
        (df["Customer Satisfaction Rating"] > 5)
    )
]

print("Invalid Satisfaction Ratings:", len(invalid_rating))

# 5. Check date columns
date_columns = [
    "Date of Purchase",
    "First Response Time",
    "Time to Resolution"
]

print("\n========== DATE CHECK ==========")

for column in date_columns:
    converted_dates = pd.to_datetime(df[column], errors="coerce")
    invalid_dates = converted_dates.isna() & df[column].notna()

    print(f"{column}:")
    print(f"  Invalid date values: {invalid_dates.sum()}")

print("\n========== CHECK COMPLETE ==========")
```

```python
# ============================================
# AUTOMATED DATA CLEANING PIPELINE
# ============================================

def clean_customer_support_data(data):
    """
    Cleans and validates the customer support ticket dataset.
    Returns a cleaned dataframe and a cleaning report.
    """

    # Create a copy so the original raw dataset remains unchanged
    cleaned = data.copy()

    # ----------------------------------------
    # 1. Remove exact duplicate rows
    # ----------------------------------------
    rows_before = len(cleaned)

    cleaned = cleaned.drop_duplicates()

    duplicate_rows_removed = rows_before - len(cleaned)

    # ----------------------------------------
    # 2. Check and handle duplicate Ticket IDs
    # ----------------------------------------
    duplicate_ticket_ids = cleaned["Ticket ID"].duplicated().sum()

    if duplicate_ticket_ids > 0:
        cleaned = cleaned.drop_duplicates(
            subset=["Ticket ID"],
            keep="first"
        )

    # ----------------------------------------
    # 3. Convert date/time columns
    # ----------------------------------------
    date_columns = [
        "Date of Purchase",
        "First Response Time",
        "Time to Resolution"
    ]

    invalid_date_counts = {}

    for column in date_columns:

        original_non_null = cleaned[column].notna()

        cleaned[column] = pd.to_datetime(
            cleaned[column],
            errors="coerce"
        )

        invalid_date_counts[column] = (
            original_non_null & cleaned[column].isna()
        ).sum()

    # ----------------------------------------
    # 4. Standardize text columns
    # ----------------------------------------
    text_columns = [
        "Customer Name",
        "Customer Email",
        "Customer Gender",
        "Product Purchased",
        "Ticket Type",
        "Ticket Subject",
        "Ticket Description",
        "Ticket Status",
        "Resolution",
        "Ticket Priority",
        "Ticket Channel"
    ]

    for column in text_columns:

        # Remove unnecessary spaces
        cleaned[column] = cleaned[column].apply(
            lambda x: x.strip() if isinstance(x, str) else x
        )

    # ----------------------------------------
    # 5. Standardize selected categorical values
    # ----------------------------------------
    categorical_columns = [
        "Customer Gender",
        "Ticket Type",
        "Ticket Status",
        "Ticket Priority",
        "Ticket Channel"
    ]

    for column in categorical_columns:

        cleaned[column] = cleaned[column].apply(
            lambda x: x.title() if isinstance(x, str) else x
        )

    # ----------------------------------------
    # 6. Validate Customer Age
    # ----------------------------------------
    invalid_age_mask = (
        (cleaned["Customer Age"] < 18) |
        (cleaned["Customer Age"] > 100)
    )

    invalid_age_count = invalid_age_mask.sum()

    # Replace invalid ages with missing values
    cleaned.loc[invalid_age_mask, "Customer Age"] = np.nan

    # ----------------------------------------
    # 7. Validate Satisfaction Rating
    # ----------------------------------------
    invalid_rating_mask = (
        cleaned["Customer Satisfaction Rating"].notna()
        &
        (
            (cleaned["Customer Satisfaction Rating"] < 1) |
            (cleaned["Customer Satisfaction Rating"] > 5)
        )
    )

    invalid_rating_count = invalid_rating_mask.sum()

    # Replace invalid ratings with missing values
    cleaned.loc[
        invalid_rating_mask,
        "Customer Satisfaction Rating"
    ] = np.nan

    # ----------------------------------------
    # 8. Generate cleaning report
    # ----------------------------------------
    cleaning_report = {
        "Rows Before Cleaning": rows_before,
        "Rows After Cleaning": len(cleaned),
        "Duplicate Rows Removed": duplicate_rows_removed,
        "Duplicate Ticket IDs Found": duplicate_ticket_ids,
        "Invalid Ages Found": int(invalid_age_count),
        "Invalid Ratings Found": int(invalid_rating_count),
        "Invalid Date Values": invalid_date_counts,
        "Missing Values After Cleaning": int(
            cleaned.isnull().sum().sum()
        )
    }

    return cleaned, cleaning_report


# ============================================
# RUN THE CLEANING PIPELINE
# ============================================

cleaned_df, cleaning_report = clean_customer_support_data(df)

print("Automated cleaning completed successfully!\n")

print("========== CLEANING REPORT ==========")

for key, value in cleaning_report.items():
    print(f"{key}: {value}")
```

```python
# ============================================
# DATA TRANSFORMATION PIPELINE
# ============================================

analysis_df = cleaned_df.copy()

# --------------------------------------------
# 1. Create Purchase Year and Month
# --------------------------------------------

analysis_df["Purchase Year"] = (
    analysis_df["Date of Purchase"].dt.year
)

analysis_df["Purchase Month"] = (
    analysis_df["Date of Purchase"].dt.month
)

analysis_df["Purchase Month Name"] = (
    analysis_df["Date of Purchase"].dt.month_name()
)

# --------------------------------------------
# 2. Calculate Response Time in Hours
# --------------------------------------------
# First Response Time is compared with Date of Purchase

analysis_df["Response Time Hours"] = (
    analysis_df["First Response Time"]
    - analysis_df["Date of Purchase"]
).dt.total_seconds() / 3600

# --------------------------------------------
# 3. Calculate Resolution Time in Hours
# --------------------------------------------
# Time to Resolution is compared with First Response Time

analysis_df["Resolution Time Hours"] = (
    analysis_df["Time to Resolution"]
    - analysis_df["First Response Time"]
).dt.total_seconds() / 3600

# --------------------------------------------
# 4. Flag resolved tickets
# --------------------------------------------

analysis_df["Is Resolved"] = (
    analysis_df["Ticket Status"].str.lower() == "closed"
)

# --------------------------------------------
# 5. Flag high-priority tickets
# --------------------------------------------

analysis_df["Is High Priority"] = (
    analysis_df["Ticket Priority"]
    .str.lower()
    .isin(["high", "critical"])
)

# --------------------------------------------
# 6. Satisfaction category
# --------------------------------------------

def satisfaction_category(rating):
    if pd.isna(rating):
        return "Not Rated"
    elif rating >= 4:
        return "Positive"
    elif rating == 3:
        return "Neutral"
    else:
        return "Negative"

analysis_df["Satisfaction Category"] = (
    analysis_df["Customer Satisfaction Rating"]
    .apply(satisfaction_category)
)

print("Data transformation completed successfully!")

print("\nNew analytical columns:")
new_columns = [
    "Purchase Year",
    "Purchase Month",
    "Purchase Month Name",
    "Response Time Hours",
    "Resolution Time Hours",
    "Is Resolved",
    "Is High Priority",
    "Satisfaction Category"
]

for column in new_columns:
    print("-", column)
```

```python
# ============================================
# STEP 10 - AUTOMATED EXCEL REPORT
# ============================================

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.chart import BarChart, LineChart, PieChart, Reference
from openpyxl.utils import get_column_letter

# Output file
excel_path = "/content/Customer_Support_Automated_Report.xlsx"

# Create workbook
wb = Workbook()

# Remove default sheet
default_sheet = wb.active
wb.remove(default_sheet)

# ============================================
# CREATE SHEETS
# ============================================

ws_summary = wb.create_sheet("Executive Summary")
ws_kpi = wb.create_sheet("KPI Summary")
ws_ticket = wb.create_sheet("Ticket Analysis")
ws_satisfaction = wb.create_sheet("Satisfaction")
ws_product = wb.create_sheet("Product Analysis")
ws_quality = wb.create_sheet("Data Quality")
ws_cleaned = wb.create_sheet("Cleaned Data")

# ============================================
# STYLES
# ============================================

header_font = Font(bold=True, color="FFFFFF")
title_font = Font(bold=True, size=16)

header_fill = PatternFill(
    "solid",
    fgColor="1F4E78"
)

subheader_fill = PatternFill(
    "solid",
    fgColor="D9EAF7"
)

thin_border = Border(
    left=Side(style="thin"),
    right=Side(style="thin"),
    top=Side(style="thin"),
    bottom=Side(style="thin")
)

# ============================================
# HELPER FUNCTION
# ============================================

def format_header(ws, row, start_col, end_col):

    for col in range(start_col, end_col + 1):

        cell = ws.cell(row=row, column=col)

        cell.font = header_font
        cell.fill = header_fill
        cell.alignment = Alignment(
            horizontal="center"
        )
        cell.border = thin_border


def autofit_columns(ws):

    for column_cells in ws.columns:

        max_length = 0
        column_letter = get_column_letter(
            column_cells[0].column
        )

        for cell in column_cells:

            try:
                length = len(str(cell.value))

                if length > max_length:
                    max_length = length

            except:
                pass

        ws.column_dimensions[
            column_letter
        ].width = min(max_length + 2, 40)


# ============================================
# EXECUTIVE SUMMARY
# ============================================

ws_summary["A1"] = "CUSTOMER SUPPORT AUTOMATED REPORT"
ws_summary["A1"].font = title_font

ws_summary["A3"] = "Project"
ws_summary["B3"] = "Data Cleaning & Reporting Automation"

ws_summary["A4"] = "Dataset"
ws_summary["B4"] = "Customer Support Ticket Dataset"

ws_summary["A6"] = "Total Tickets"
ws_summary["B6"] = total_tickets

ws_summary["A7"] = "Closed Tickets"
ws_summary["B7"] = closed_tickets

ws_summary["A8"] = "Open Tickets"
ws_summary["B8"] = open_tickets

ws_summary["A9"] = "Pending Tickets"
ws_summary["B9"] = pending_tickets

ws_summary["A10"] = "High/Critical Tickets"
ws_summary["B10"] = high_priority_tickets

ws_summary["A11"] = "Average Satisfaction"
ws_summary["B11"] = round(
    average_satisfaction, 2
)

ws_summary["A12"] = "Resolution Rate (%)"
ws_summary["B12"] = round(
    resolution_rate, 2
)

ws_summary["A13"] = "Positive Satisfaction Rate (%)"
ws_summary["B13"] = round(
    positive_satisfaction_rate, 2
)

for row in range(6, 14):

    ws_summary[f"A{row}"].font = Font(
        bold=True
    )

    ws_summary[f"A{row}"].border = thin_border
    ws_summary[f"B{row}"].border = thin_border

# ============================================
# KPI SUMMARY
# ============================================

ws_kpi.append(["KPI", "Value"])

for _, row in kpi_summary.iterrows():

    ws_kpi.append([
        row["KPI"],
        row["Value"]
    ])

format_header(
    ws_kpi,
    1,
    1,
    2
)

# ============================================
# TICKET ANALYSIS
# ============================================

ws_ticket["A1"] = "TICKET ANALYSIS"
ws_ticket["A1"].font = title_font

# Ticket Type
ws_ticket["A3"] = "Tickets by Type"
ws_ticket["A3"].font = Font(bold=True)

row_number = 4

ws_ticket.cell(
    row=row_number,
    column=1,
    value="Ticket Type"
)

ws_ticket.cell(
    row=row_number,
    column=2,
    value="Ticket Count"
)

format_header(
    ws_ticket,
    row_number,
    1,
    2
)

row_number += 1

for _, row in tickets_by_type.iterrows():

    ws_ticket.cell(
        row=row_number,
        column=1,
        value=row["Ticket Type"]
    )

    ws_ticket.cell(
        row=row_number,
        column=2,
        value=row["Ticket Count"]
    )

    row_number += 1

# Ticket Status
status_start = row_number + 2

ws_ticket.cell(
    row=status_start,
    column=1,
    value="Tickets by Status"
)

ws_ticket.cell(
    row=status_start + 1,
    column=1,
    value="Ticket Status"
)

ws_ticket.cell(
    row=status_start + 1,
    column=2,
    value="Ticket Count"
)

format_header(
    ws_ticket,
    status_start + 1,
    1,
    2
)

r = status_start + 2

for _, row in tickets_by_status.iterrows():

    ws_ticket.cell(
        row=r,
        column=1,
        value=row["Ticket Status"]
    )

    ws_ticket.cell(
        row=r,
        column=2,
        value=row["Ticket Count"]
    )

    r += 1

# Priority
priority_start = r + 2

ws_ticket.cell(
    row=priority_start,
    column=1,
    value="Tickets by Priority"
)

ws_ticket.cell(
    row=priority_start + 1,
    column=1,
    value="Ticket Priority"
)

ws_ticket.cell(
    row=priority_start + 1,
    column=2,
    value="Ticket Count"
)

format_header(
    ws_ticket,
    priority_start + 1,
    1,
    2
)

r = priority_start + 2

for _, row in tickets_by_priority.iterrows():

    ws_ticket.cell(
        row=r,
        column=1,
        value=row["Ticket Priority"]
    )

    ws_ticket.cell(
        row=r,
        column=2,
        value=row["Ticket Count"]
    )

    r += 1

# Channel
channel_start = r + 2

ws_ticket.cell(
    row=channel_start,
    column=1,
    value="Tickets by Channel"
)

ws_ticket.cell(
    row=channel_start + 1,
    column=1,
    value="Ticket Channel"
)

ws_ticket.cell(
    row=channel_start + 1,
    column=2,
    value="Ticket Count"
)

format_header(
    ws_ticket,
    channel_start + 1,
    1,
    2
)

r = channel_start + 2

for _, row in tickets_by_channel.iterrows():

    ws_ticket.cell(
        row=r,
        column=1,
        value=row["Ticket Channel"]
    )

    ws_ticket.cell(
        row=r,
        column=2,
        value=row["Ticket Count"]
    )

    r += 1

# ============================================
# SATISFACTION SHEET
# ============================================

ws_satisfaction["A1"] = "CUSTOMER SATISFACTION"
ws_satisfaction["A1"].font = title_font

ws_satisfaction.append([])
ws_satisfaction.append([
    "Satisfaction Category",
    "Ticket Count"
])

format_header(
    ws_satisfaction,
    3,
    1,
    2
)

for _, row in satisfaction_summary.iterrows():

    ws_satisfaction.append([
        row["Satisfaction Category"],
        row["Ticket Count"]
    ])

# ============================================
# PRODUCT ANALYSIS
# ============================================

ws_product["A1"] = "PRODUCT ANALYSIS"
ws_product["A1"].font = title_font

ws_product.append([])
ws_product.append([
    "Product",
    "Ticket Count"
])

format_header(
    ws_product,
    3,
    1,
    2
)

for _, row in tickets_by_product.iterrows():

    ws_product.append([
        row["Product"],
        row["Ticket Count"]
    ])

# ============================================
# DATA QUALITY
# ============================================

ws_quality["A1"] = "DATA QUALITY REPORT"
ws_quality["A1"].font = title_font

quality_data = [
    ("Rows Before Cleaning", cleaning_report["Rows Before Cleaning"]),
    ("Rows After Cleaning", cleaning_report["Rows After Cleaning"]),
    ("Duplicate Rows Removed", cleaning_report["Duplicate Rows Removed"]),
    ("Duplicate Ticket IDs Found", cleaning_report["Duplicate Ticket IDs Found"]),
    ("Invalid Ages Found", cleaning_report["Invalid Ages Found"]),
    ("Invalid Ratings Found", cleaning_report["Invalid Ratings Found"]),
    ("Missing Values After Cleaning", cleaning_report["Missing Values After Cleaning"]),
    ("Invalid Date Values - Purchase", 0),
    ("Invalid Date Values - Response", 0),
    ("Invalid Date Values - Resolution", 0),
    ("Negative Resolution Durations", int(negative_resolution)),
    ("Response Durations > 7 Days", int(large_response))
]

ws_quality.append([])

for item, value in quality_data:

    ws_quality.append([
        item,
        value
    ])

# ============================================
# CLEANED DATA
# ============================================

# Write column names
for col_num, column_name in enumerate(
    analysis_df.columns,
    start=1
):

    ws_cleaned.cell(
        row=1,
        column=col_num,
        value=column_name
    )

format_header(
    ws_cleaned,
    1,
    1,
    len(analysis_df.columns)
)

# Write data
for row_num, row in enumerate(
    analysis_df.itertuples(index=False),
    start=2
):

    for col_num, value in enumerate(
        row,
        start=1
    ):

        # Convert pandas NaT/NaN to blank
        if pd.isna(value):
            value = None

        ws_cleaned.cell(
            row=row_num,
            column=col_num,
            value=value
        )

# Freeze header
ws_cleaned.freeze_panes = "A2"

# ============================================
# ADD BASIC CHARTS
# ============================================

# Status chart
chart = BarChart()

chart.title = "Tickets by Status"
chart.y_axis.title = "Ticket Count"
chart.x_axis.title = "Status"

data = Reference(
    ws_ticket,
    min_col=2,
    min_row=status_start + 1,
    max_row=status_start + 4
)

categories = Reference(
    ws_ticket,
    min_col=1,
    min_row=status_start + 2,
    max_row=status_start + 4
)

chart.add_data(
    data,
    titles_from_data=True
)

chart.set_categories(categories)

ws_ticket.add_chart(
    chart,
    "D3"
)

# Priority chart
priority_chart = BarChart()

priority_chart.title = "Tickets by Priority"

data = Reference(
    ws_ticket,
    min_col=2,
    min_row=priority_start + 1,
    max_row=priority_start + 5
)

categories = Reference(
    ws_ticket,
    min_col=1,
    min_row=priority_start + 2,
    max_row=priority_start + 5
)

priority_chart.add_data(
    data,
    titles_from_data=True
)

priority_chart.set_categories(
    categories
)

ws_ticket.add_chart(
    priority_chart,
    "D20"
)

# ============================================
# FORMAT ALL SHEETS
# ============================================

for worksheet in wb.worksheets:

    worksheet.freeze_panes = (
        worksheet.freeze_panes or "A2"
    )

    autofit_columns(worksheet)

# ============================================
# SAVE REPORT
# ============================================

wb.save(excel_path)

print("============================================")
print("EXCEL REPORT GENERATED SUCCESSFULLY!")
print("============================================")
print(f"File: {excel_path}")
print("\nSheets created:")

for sheet in wb.sheetnames:
    print("-", sheet)
```
